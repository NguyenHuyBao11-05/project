```python
def check_triangle(a,b,c):
  if ( a + b <= c ) or ( a + c <= b ) or ( b + c <= a) :
    print ("Khong Phai La Tam Giac")
  else :
      if ( a == b ) and ( a == c ):
        print ("Tam Giac Deu")
      elif ( a == b ) or ( a == c ) or ( b == c ):
        print ("Tam Giac Can")
      else :
        print ("Tam Giac Thuong")

```

```python
check_triangle(5,5,3)
```

```python
def check_point_rectangle(x,y,w,h) :
  x1 = int(input("Nhap toa do x1: "))
  y1 = int(input("Nhap toa do y1: "))

  if (w <= 0) and (h <= 0) :
    print ("Khong phai la hinh chu nhat")

  x_inside = x <= x + x1 <= x+w
  y_inside = y <= y + y1 <= y+h

  if x_inside and y_inside :
    print (x1,y1, "thuoc hinh chu nhat")
  else :
    print (x1,y1, "khong thuoc hinh chu nhat")

```

```python
check_point_rectangle(4,4,3,4)
```

```python
def findM(n):
  if n < 1 :
    return 0
  m = 0
  sum = 0
  while (sum+m+1 < n) :
    m += 1
    sum += m
  return m

```

```python
findM(11)
```

```python
def fibonaci(num):
  if num == 0:
    return 0
  elif num == 1 :
    return 1
  else :
    return fibonaci(num-1) + fibonaci(num-2)
```

```python
fibonaci(8)
```
